```python
import pandas as pd

df = pd.read_csv("/content/PAMAP2_Dataset/Protocol/subject101.dat", sep=' ', header=None)
```

```python
df = df[[1, 2]]  # 1 = activity_id, 2 = heart_rate
df.columns = ['activity_id', 'heart_rate']

df = df.dropna()
```

```python
activity_map = {
    1: "lying",
    2: "sitting",
    3: "standing",
    4: "walking",
    5: "running",
    6: "cycling"
}

df['activity'] = df['activity_id'].map(activity_map)

# Sirf useful activities rakho
df = df[df['activity'].isin(["walking", "running", "cycling"])]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X = df[['heart_rate']]
y = df['activity']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestClassifier()
model.fit(X_train, y_train)

print("Accuracy:", model.score(X_test, y_test))
```

```python
import json
import os

exercise_path = "/content/free-exercise-db-main/exercises"

exercises = []

for file in os.listdir(exercise_path):
    if file.endswith(".json"):   # sirf JSON files

        with open(os.path.join(exercise_path, file)) as f:
            data = json.load(f)

            # kuch files list hoti hain, kuch dict
            if isinstance(data, list):
                exercises.extend(data)
            else:
                exercises.append(data)

print("Total exercises:", len(exercises))
print(exercises[0])


```

```python
def get_exercises_by_type(ex_type):

    if ex_type == "cardio":
        return [
            ex for ex in exercises
            if (
                # ✔ include
                ("jump" in ex.get('name', '').lower()
                 or "run" in ex.get('name', '').lower()
                 or "skip" in ex.get('name', '').lower()
                 or "sprint" in ex.get('name', '').lower())

                # ❌ exclude unwanted
                and "crunch" not in ex.get('name', '').lower()
                and "stretch" not in ex.get('name', '').lower()
                and "press" not in ex.get('name', '').lower()
                and "push" not in ex.get('name', '').lower()
                and "curl" not in ex.get('name', '').lower()
                and "weighted" not in ex.get('name', '').lower()
            )
        ]

    elif ex_type == "light":

        # 🔹 Light movement (NO heavy equipment)
        movement = [
            ex for ex in exercises
            if (
                ("walk" in ex.get('name', '').lower()
                 or "step" in ex.get('name', '').lower()
                 or "lunge" in ex.get('name', '').lower())
                and ex.get('equipment') in ['body only', None]
            )
        ]

        # 🔹 Lower body stretch
        stretch = [
            ex for ex in exercises
            if "stretch" in ex.get('name', '').lower()
            and (
                "leg" in ex.get('name', '').lower()
                or "hamstring" in ex.get('name', '').lower()
                or "quad" in ex.get('name', '').lower()
            )
        ]

        return movement[:3] + stretch[:2]

    else:
        return [
            ex for ex in exercises
            if ex.get('category') == 'strength'
            and ex.get('level') == 'beginner'
        ]
```

```python
def map_activity_to_exercise(activity):

    if activity == "running":
        return "cardio"

    elif activity == "walking":
        return "light"

    elif activity == "cycling":
        return "cardio"

    else:
        return "strength"
```

```python
def recommend_exercises(activity):

    ex_type = map_activity_to_exercise(activity)

    filtered = get_exercises_by_type(ex_type)

    import random
    random.shuffle(filtered)

    # 🔥 remove duplicates
    seen = set()
    unique = []

    for ex in filtered:
        name = ex['name']
        if name not in seen:
            unique.append(ex)
            seen.add(name)

    # 🔥 pick 5
    selected = unique[:5]

    return [ex['name'] for ex in selected]
```

```python
def predict_and_recommend(heart_rate):

    import pandas as pd

    # 🔥 Rule-based override
    if heart_rate < 100:
        activity = "walking"

    elif heart_rate < 130:
        activity = "cycling"

    else:
        activity = "running"

    recs = recommend_exercises(activity)

    return activity, recs
```

```python
activity, recs = predict_and_recommend(100)

print("Activity:", activity)

for r in recs:
    print("-", r)
```

```python
test_values = [70, 90, 110, 130, 150]

for hr in test_values:
    activity, recs = predict_and_recommend(hr)

    print("\nHR:", hr)
    print("Activity:", activity)

    for r in recs:
        print("-", r)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X = df[['heart_rate']]
y = df['activity']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestClassifier()
model.fit(X_train, y_train)

print("Accuracy:", model.score(X_test, y_test))
```

```python
from flask import Flask, request, jsonify
import pandas as pd

app = Flask(__name__)

@app.route("/predict", methods=["POST"])
def predict():

    data = request.json
    heart_rate = data.get("heart_rate", 0)

    activity, recs = predict_and_recommend(heart_rate)

    return jsonify({
        "activity": activity,
        "exercises": recs
    })

app.run(host="0.0.0.0", port=5000)
```
